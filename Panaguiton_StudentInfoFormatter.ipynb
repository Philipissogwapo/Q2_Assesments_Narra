```python
while True:
  first_name = input("\nWhat is your name? : ")
  if first_name == "" or first_name == " ":
      print("\nCannot leave it empty, try again.")
      continue
  else:
     break

while True:
  last_name = input("\nWhat is your last name? : ")
  if last_name == "" or last_name == " ":
      print("\nCannot leave it empty, try again.")
      continue
  else:
     break

while True:
  section = input("\nWhat is your section? : ")
  if section == "" or section == " ":
      print("\nCannot leave it empty, try again.")
      continue
  else:
     break

while True:
  favorite_subject = input("\nWhat is your favorite subject? : ")
  if favorite_subject == "" or favorite_subject == " ":
      print("\nCannot leave it empty, try again.")
      continue
  else:
     break

first_name_formatted = first_name.strip().title()
last_name_formatted = last_name.strip().title()
first_name_formatted_capital = first_name.strip().upper()
last_name_formatted_capital = last_name.strip().upper()
section_formatted = section.strip().upper()
favorite_subject_formatted = favorite_subject.strip().title()
character_count = len(first_name_formatted) + len(last_name_formatted) + 1

print("\n---Formatted Output---")
print(f"\nName : {first_name_formatted} {last_name_formatted}")
print(f"\nName Code : {first_name_formatted_capital} {last_name_formatted_capital}")
print(f"\nSection : {section_formatted}")
print(f"\nFavorite Subject : {favorite_subject_formatted}")
print(f"\nCharacter Count : {character_count}")
```
