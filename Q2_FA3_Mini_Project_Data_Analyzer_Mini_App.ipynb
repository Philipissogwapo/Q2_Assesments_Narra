```python
favorite_foods = []

print("=== Favorite Foods ===")

for i in range(3):
  favorite_food = input("\nPlease input 3 of your favorite foods : ")
  str(favorite_food)

  favorite_foods.append(favorite_food)

while True:
    user_interface = int(input("\nWhat would you like to do next?\n\n1 - Display all favorite foods\n2 - Add a favorite food\n3 - Remove a favorite food\n4 - Exit\n\nPlease input the number you want to do : "))

    if user_interface == 1:
      for i in range(len(favorite_foods)):
        print("\n", favorite_foods[i])
```
