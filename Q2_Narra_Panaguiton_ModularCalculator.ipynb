```python
def add_numbers(num1, num2):
  return num1 + num2

def subtract_numbers(num1, num2):
  return num1 - num2

def multiply_numbers(num1, num2):
  return round(num1 * num2, 2)

def divide_numbers(num1, num2):
   if num2 == 0:
     return("Cannot divide by 0")
   else:
     quotient = num1 / num2
     rounded_quotient = round(quotient, 2)
     return rounded_quotient
# These are the functions to the operations

print("---Modular Calculator---\n")
print("To use this calculator, enter 2 numbers, select an operation, and the calculator will give a result.\n")
# Introduces them to the calculator

while True:
  try:
    num1 = float(input("Please input your first number : "))
    break
  except ValueError:
    print("\nCannot put that value, try again.\n")
# Asks for number 1

while True:
  try:
    num2 = float(input("\nPlease input your second number : "))
    break
  except ValueError:
    print("\nCannot put that value, try again.\n")
# Asks for number 2

while True:
  try:
    operation = int(input("\n1 - Addition\n2 - Subtraction\n3 - Multiplication\n4 - Division\n\nPlease input the operation to be used : "))

    if 1 > operation or operation > 4:
      print("\nYou cannot input numbers less than 1 or greater than 4. Please try again.")
      continue
    else:
      break
  except ValueError:
    print("\nNumbers 1 - 4 only, nothing else. Please try again.")
# Makes them choose an operation

if operation == 1:
  addition = add_numbers(num1, num2)
  print(f"\nThe answer is : {addition}.\n")
elif operation == 2:
  subtraction = subtract_numbers(num1, num2)
  print(f"\nThe answer is : {subtraction}.\n")
elif operation == 3:
  multiplication = multiply_numbers(num1, num2)
  print(f"\nThe answer is : {multiplication}.\n")
else:
  division = divide_numbers(num1, num2)
  print(f"\nThe answer is : {division}.\n")
# Displays the answer

print("---End---")
```

What functions did you create in your calculator program?
- I created functions for addition, subtraction, multiplication, and division.

What parameters did your functions use?
- I used num1 and num2 as the parameters for my functions.

What arguments were passed when the functions were called?
- The arguments used are the inputs for the num1 and num2 (global).

How did your program use the returned value?
- I saved the return value into variables so that they can be used later when displaying the answer.

Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
- It's better to divide the program into smaller parts so that if an error happens, you can spot the part that made the error and fix it faster.
